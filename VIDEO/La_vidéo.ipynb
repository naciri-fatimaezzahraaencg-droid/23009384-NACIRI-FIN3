```python
from IPython.display import HTML
from base64 import b64encode

# Read the video file and encode it
with open('/content/F5_Facteurs_de_risque_portefeuille.mp4', 'rb') as f:
    video_bytes = f.read()

video_encoded = b64encode(video_bytes).decode('utf-8')
video_url = f'data:video/mp4;base64,{video_encoded}'

# Display the video with play controls
HTML(f"""
<video width="640" height="480" controls>
  <source src="{video_url}" type="video/mp4">
  Your browser does not support the video tag.
</video>
""")
```
